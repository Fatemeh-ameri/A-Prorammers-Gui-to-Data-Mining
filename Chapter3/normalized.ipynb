```python
userRating = {"David": {"Imagine Dragons": 3, "Daft Punk":5,"Lorde": 4, "Fall Out Boy": 1},
"Matt": {"Imagine Dragons": 3, "Daft Punk": 4,"Lorde": 4, "Fall Out Boy": 1},
"Ben": {"Kacey Musgraves": 4, "Imagine Dragons": 3,"Lorde": 3, "Fall Out Boy": 1},
"Chris": {"Kacey Musgraves": 4, "Imagine Dragons": 4,"Daft Punk": 4, "Lorde": 3, "Fall Out Boy": 1}, 
"Tori": {"Kacey Musgraves": 5, "Imagine Dragons": 4,"Daft Punk": 5, "Fall Out Boy": 3}}
```

```python
def normalized(input, userRating, min =1, max =5):
    normalizedRatings = {}
    for movie in userRating[input]:
        normal1 = 2 * (userRating[input][movie] - min) - (max-min)
        normal2 = max - min
        normal = normal1 / normal2

        normalizedRatings[movie] = normal
    return normalizedRatings
```

```python
davidNormalizedRatings = normalized('David', userRating)
print(davidNormalizedRatings)
```

```python
def denormalized(input, min=1, max=5):
    denormalizedRatings = {}
    for movie in input:
        normal = (((input[movie] + 1) * (max - min)) / 2) + min
        denormalizedRatings[movie] = normal
    return denormalizedRatings
```

```python
davidDenormalizedRatings = denormalized(davidNormalizedRatings)
print(davidDenormalizedRatings)
```
