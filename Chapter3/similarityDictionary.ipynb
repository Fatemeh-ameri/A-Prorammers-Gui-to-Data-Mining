```python
import numpy as np
from math import sqrt

# Define the user ratings dictionary
userRating = {
    "David": {"Imagine Dragons": 3, "Daft Punk": 5, "Lorde": 4, "Fall Out Boy": 1},
    "Matt": {"Imagine Dragons": 3, "Daft Punk": 4, "Lorde": 4, "Fall Out Boy": 1},
    "Ben": {"Kacey Musgraves": 4, "Imagine Dragons": 3, "Lorde": 3, "Fall Out Boy": 1},
    "Chris": {"Kacey Musgraves": 4, "Imagine Dragons": 4, "Daft Punk": 4, "Lorde": 3, "Fall Out Boy": 1},
    "Tori": {"Kacey Musgraves": 5, "Imagine Dragons": 4, "Daft Punk": 5, "Fall Out Boy": 3}
}

# Step 1: Calculate the average rating for each user
averages = {}
for user, ratings in userRating.items():
    averages[user] = sum(ratings.values()) / len(ratings.values())

# Step 2: Adjusted Cosine Similarity function
def adjustedCosineSimilarity(band1, band2, userRating):
    numerator = 0
    denominator1 = 0
    denominator2 = 0
    
    for user, ratings in userRating.items():
        if band1 in ratings and band2 in ratings:
            avg = averages[user]
            numerator += (ratings[band1] - avg) * (ratings[band2] - avg)
            denominator1 += (ratings[band1] - avg) ** 2
            denominator2 += (ratings[band2] - avg) ** 2
    
    if denominator1 == 0 or denominator2 == 0:
        return 0
    
    return numerator / (sqrt(denominator1) * sqrt(denominator2))

# Step 3: Create the similarity dictionary
bands = list({band for ratings in userRating.values() for band in ratings})
similarity_dict = {}

for i in range(len(bands)):
    similarity_dict[bands[i]] = {}
    for j in range(len(bands)):
        if i != j:
            similarity_dict[bands[i]][bands[j]] = adjustedCosineSimilarity(bands[i], bands[j], userRating)

# Print the similarity dictionary
print("Music Band Similarity Dictionary:")
for band, similarities in similarity_dict.items():
    print(f"{band}: {similarities}")
```
