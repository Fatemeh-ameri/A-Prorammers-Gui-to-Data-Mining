```python
from math import sqrt

# Define the user ratings dictionary
userRating = {
    "David": {"Imagine Dragons": 3, "Daft Punk": 5, "Lorde": 4, "Fall Out Boy": 1},
    "Matt": {"Imagine Dragons": 3, "Daft Punk": 4, "Lorde": 4, "Fall Out Boy": 1},
    "Ben": {"Kacey Musgraves": 4, "Imagine Dragons": 3, "Lorde": 3, "Fall Out Boy": 1},
    "Chris": {"Kacey Musgraves": 4, "Imagine Dragons": 4, "Daft Punk": 4, "Lorde": 3, "Fall Out Boy": 1},
    "Tori": {"Kacey Musgraves": 5, "Imagine Dragons": 4, "Daft Punk": 5, "Fall Out Boy": 3}
}

# Step 1: Normalize Dave's ratings
def normalized(input, userRating, min=1, max=5):
    normalizedRatings = {}
    for movie in userRating[input]:
        normal1 = 2 * (userRating[input][movie] - min) - (max - min)
        normal2 = max - min
        normal = normal1 / normal2
        normalizedRatings[movie] = normal
    return normalizedRatings

# Step 2: Denormalize ratings
def denormalized(input, min=1, max=5):
    denormalizedRatings = {}
    for movie in input:
        normal = (((input[movie] + 1) * (max - min)) / 2) + min
        denormalizedRatings[movie] = normal
    return denormalizedRatings

# Step 3: Adjusted Cosine Similarity function
def adjustedCosineSimilarity(band1, band2, userRating):
    averages = {user: sum(ratings.values()) / len(ratings) for user, ratings in userRating.items()}
    numerator = 0
    denominator1 = 0
    denominator2 = 0
    
    for user, ratings in userRating.items():
        if band1 in ratings and band2 in ratings:
            avg = averages[user]
            numerator += (ratings[band1] - avg) * (ratings[band2] - avg)
            denominator1 += (ratings[band1] - avg) ** 2
            denominator2 += (ratings[band2] - avg) ** 2
    
    if denominator1 == 0 or denominator2 == 0:
        return 0
    
    return numerator / (sqrt(denominator1) * sqrt(denominator2))

# Normalize Dave's ratings
dave_normalized_ratings = normalized("David", userRating)

# Calculate the similarity between "Kacey Musgraves" and each of the bands Dave has rated
bands_dave_rated = userRating["David"].keys()
similarities = {}
for band in bands_dave_rated:
    similarities[band] = adjustedCosineSimilarity("Kacey Musgraves", band, userRating)

# Predict Dave's normalized rating for "Kacey Musgraves"
numerator = 0
denominator = 0
for band, similarity in similarities.items():
    numerator += similarity * dave_normalized_ratings[band]
    denominator += abs(similarity)

# Handle division by zero in case no similar items are found
if denominator != 0:
    predicted_normalized_rating = numerator / denominator
else:
    predicted_normalized_rating = 0  # default to zero if no valid similarities

# Denormalize the predicted rating
predicted_normalized_dict = {"Kacey Musgraves": predicted_normalized_rating}
predicted_denormalized_rating = denormalized(predicted_normalized_dict)["Kacey Musgraves"]

print(f"Predicted rating for Dave on Kacey Musgraves: {predicted_denormalized_rating:.3f}")
```
